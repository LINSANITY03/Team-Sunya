```python
crop_depth_data_path = Path().resolve().parent / "data" / "Crop" / "raw" / "Crop_Rooting_Depth_Data.xlsx"
crop_depth = pd.read_excel(crop_depth_data_path)
crop_depth
```

```python
# Convert depth DataFrame into a lookup dictionary
depth_dict = dict(zip(crop_depth['Crop'], crop_depth['Rooting depth (m)']))
depth_dict
```

```python
explicit_mapping = {
    'Rice / Paddy': 'Rice',
    'Soybean': 'Soybeans',
    'Maize / Corn': 'Maize',
    'Upland Rice': 'Rice',
}

def get_category_fallback(crop_name):
    name = crop_name.lower()
    
    if any(k in name for k in ['lentil', 'chickpea', 'pigeonpea', 'gram', 'mungbean', 'pulse']):
        return depth_dict.get('Pulses', 0.6)
    if any(k in name for k in ['rapeseed', 'mustard', 'linseed', 'oil']):
        return depth_dict.get('Other oil crops', 1.0)
    if any(k in name for k in ['buckwheat', 'cereal', 'oat', 'barley', 'millet']):
        return depth_dict.get('Other cereals', 1.0)
    if any(k in name for k in ['root', 'tuber', 'potato']):
        return depth_dict.get('Other roots', 0.5)
    if any(k in name for k in ['fiber', 'jute', 'dhaincha', 'cotton']):
        return depth_dict.get('Fibres', 0.8)
        
    # Ultimate default if no category matches
    return 'N/A'

def get_rooting_depth(crop_name):
    # Tier A: Check explicit dictionary mapping first
    target_name = explicit_mapping.get(crop_name, crop_name)
    if target_name in depth_dict:
        return depth_dict[target_name]
    
    # Tier B: Check direct match in depth dict
    if crop_name in depth_dict:
        return depth_dict[crop_name]
        
    # Tier C: Apply automated keyword category rules for unmapped crops
    return get_category_fallback(crop_name)

# Fix: Apply and store the result in 'knowledge_df'
knowledge_df['Rooting depth (m)'] = knowledge_df['Common Crop Name'].apply(get_rooting_depth)

knowledge_df
```

```python
lat = 27.643692887895675
lon = 84.36065413845583

FIVE_CM = "0-5cm"
FIVETEEN_CM = "5-15cm"
THIRTY_CM = "15-30cm"
SIXTY_CM = "30-60cm"
HUNDRED_CM = "60-100cm"
TWO_HUNDRED_CM = "100-200cm"

ROOT_LEVEL = {
    0 : FIVE_CM,
    1 : FIVETEEN_CM,
    2 : THIRTY_CM,
    3 : SIXTY_CM,
    4 : HUNDRED_CM,
    5 : TWO_HUNDRED_CM,
}

def get_soil_depth(root_depth):
    root_depth_in_cm = root_depth * 100
    if root_depth_in_cm < 5:
        return 0
    elif root_depth_in_cm >= 5 and root_depth_in_cm < 15:
        return 1
    elif root_depth_in_cm >= 15 and root_depth_in_cm < 30:
        return 2
    elif root_depth_in_cm >= 30 and root_depth_in_cm < 60:
        return 3
    elif root_depth_in_cm >= 60 and root_depth_in_cm < 100:
        return 4
    elif root_depth_in_cm >= 100 and root_depth_in_cm < 200:
        return 5
    else:
        return None

EXAMPLE_CROP = "Rice"

def strict_slash_match(search_term, target_string):
    # 1. Split the target string by '/' and strip whitespace/lowercase both sides
    parts = [p.strip().lower() for p in target_string.split('/')]
    
    # 2. Check if the search term strictly matches any individual part
    return search_term.lower() in parts

# Apply the strict slash match function across the column
mask = knowledge_df["Common Crop Name"].apply(lambda x: strict_slash_match(EXAMPLE_CROP, x))

# Filter the dataframe and get the rooting depth
matching_row = knowledge_df[mask]
depth_value = matching_row["Rooting depth (m)"].iloc[0]
print(depth_value)
```

```python
required_depth = [ROOT_LEVEL[i] for i in range(get_soil_depth(depth_value)+1)]
print(required_depth)
```

```python
import numpy as np
import rasterio
from rasterio.warp import transform

def get_soil_properties(depth_range):
    return {
        "clay": {"depth": depth_range, "scale": 10, "unit": "%"},
        "sand": {"depth": depth_range, "scale": 10, "unit": "%"},
        "silt": {"depth": depth_range, "scale": 10, "unit": "%"},
    }

def get_raster_url(prop, meta):
    # Keep your existing URL for standard soil properties.
    return (
        "https://files.isric.org/soilgrids/latest/data/"
        f"{prop}/{prop}_{meta['depth']}_mean.vrt"
    )

def extract_soil_value(prop, meta, lon, lat):
    """Extract one SoilGrids raster value at a coordinate."""

    # Inside your existing extraction function:
    url = get_raster_url(prop, meta)

    with rasterio.open(url) as src:
        xs, ys = transform(
            "EPSG:4326",
            src.crs,
            [lon],
            [lat],
        )

        raw_value = next(src.sample([(xs[0], ys[0])]))[0]

        # Handle nodata and masked pixels.
        if np.ma.is_masked(raw_value):
            return None

        raw_value = float(raw_value)

        if not np.isfinite(raw_value):
            return None

        if src.nodata is not None and raw_value == src.nodata:
            return None

        if raw_value <= -9999:
            return None

        return raw_value / meta["scale"]

def extract_properies(batch_prop):
    extracted_records = []
    # Extract all properties
    for prop, meta in batch_prop.items():
        try:
            value = extract_soil_value(prop, meta, lon, lat)

            extracted_records.append({  
                "Property": prop,
                "Depth": meta["depth"],
                "Value": value,
                "Unit": meta["unit"],
            })

            print(f"{prop}: {value} {meta['unit']}")

        except Exception as exc:
            print(f"Error fetching {prop}: {exc}")

            extracted_records.append({
                "Property": prop,
                "Depth": meta["depth"],
                "Value": None,
                "Unit": meta["unit"],
            })
            
    return extracted_records

final_result = dict()

for each in required_depth:
    each_depth_range = get_soil_properties(each)
    if each in final_result:
        final_result.update(extract_properies(each_depth_range))
    else:
        final_result[each] = extract_properies(each_depth_range)

print(final_result)
```
